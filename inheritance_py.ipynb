```python
'''inheritance allow derived class to inherit the attributes and methods of parent class
that is one class gets the attributes or behaviour of another class
```

```python
#creating parent class
class animal:
    def eat(self):
        print("this animal eats food")
```

```python
#creating child class
class dog(animal):
    def bark(self):
        print("dog barks")

#object creation
d=dog()
d.eat()   #inherited from animal
d.bark()  #dog's own method
```

```python
class user:
    def login(self):
        print("user login")

class student(user):
    def watch_video(self):
        print("students are watching video")

class teacher(user):
    def upload_video(self):
        print("teachers are uploading video")

s=student()
s.login()
s.watch_video()
t=teacher()
t.login()
t.upload_video()
```

```python
# multiple inheritance (2 parent class 1 child class)
class father:
    def skill(self):
        # method of father class
        print("father's skill:gardening")

class mother:
    def skill2(self):
        # method of mother class
        print("mother's skill:cooking")

class child(father,mother):
    # child inherits from both mother and father
    pass
#creating object of child class
c=child()
#calling method from father class
c.skill()
#calling method from mother class
c.skill2()
```

```python
# multi level inheritance (Grand Parent class->parent class->child class)
class grandparent:
    def display(self):
        print("grand parent class")

class parent(grandparent):
    def show(self):
        print("parent class")

class child(parent):
    def see(self):
        print("child class")

c=child()
c.display()
c.show()
c.see()
```

```python
#hybrid inheritance (child inherits from parents and grandparents)

# base class
class animal:
    def eat(self):
        print("animal eats")

# derived class (single inheritance)
class mammal(animal):
    def walk(self):
        print("mammal walks")

#another base class
class pet:
    def play(self):
        print("pet plays")

# derived class (multiple inheritance)
class dog(mammal,pet):
    def bark(self):
        print("dog barks")

#creating object
d=dog()

#accessing method
d.eat() #from animal
d.walk() #from mammal
d.play() #from pet
d.bark() #from dog
```
