```python
#accessing the elements

my_dict={"name":"alice","age":20,"city":"New York"}
print(my_dict["name"])
```

```python
#modify the existing elements

my_dict["name"]="RIYA"
my_dict["age"]=18
my_dict
```

```python
# removing an element

del my_dict["city"]
my_dict
```

```python
#pop()

my_dict.pop("age")
my_dict
```

```python
#clear()

my_dict.clear()
my_dict
```

```python
#len()

D={"name":"alice","age":20,"city":"New York"}
len(D)
```

```python
#keys()

D.keys()
```

```python
#values()

D.values()
```

```python
#items()

D.items()
```
