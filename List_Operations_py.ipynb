```python
#adding element using append()
a=[9,8,7,6]
a.append(5)
print(a)
```

```python
#adding element using insert()
r=[1,2,3,4]
r.insert(1,22)
r
```

```python
#inserting element using extend()
r=[1,2,3,4,5]
r.extend((6,7,8,9,0))
print(r)
```

```python
#removing an element using del()
a=[10,20,30,40,50]
del a[3]
print(a)
```

```python
#removing an element from list using remove()
my_list=[10,20,30,40,50]
my_list.remove(30)
print(my_list)
```

```python
#removing an element from list using pop()
my_list=[10,20,30,40,50]
removed_element=my_list.pop()
print("removed element:",removed_element)
print("updated list:",my_list)
```

```python
#concatenation
#using + operator
list1=[1,2,3]
list2=[4,5,6]
concatenated_list=list1+list2
concatenated_list
```

```python
#concatenation
#using extend()
my_list=[1,2,3]
my_list.extend((5,6,7))
my_list
```

```python
#Repitition(*)
original_list=[9,7,5]
repeated_list=original_list*3
print(repeated_list)
```

```python
#count
my_list=[10,20,30,20,40,50,20]
my_list.count(20)
```

```python
#reverse
R=[0,9,8,7,6,5,4,3,2,1]
R.reverse()
print("reversed list=",R)
```

```python
#sort
my_list=[3,1,4,1,5,9,2,6]
my_list.sort()
print(my_list)
my_list.sort(reverse=True)
print(my_list)
words=["banana","apple","cherry"]
words.sort(key=len)
print(words)
```
