```python
#exception handling in python is used to manage runtime errors
try:
    a=int(input("Enter a number:"))
    print(10/a)
except:
    print("Error occured")
```

```python
try:
    a=int(input("Enter a number:"))
    print(10/a)
except:
    print("Error occured")
```

```python
a=4
b=0
try:
    print(a/b)
except:
    print("some kind of error")
else:
    print("no error")
finally:
    print("bye")
```

```python
try:
    balance=5000
    amount=int(input("enter withdrawel amount:"))
    if amount>balance:
        raise ValueError("insufficient balance")
    print("please collect your cash")
except ValueError as e:
    print("transaction failed:",e)
finally:
    print("thank you for using our ATM")
```
