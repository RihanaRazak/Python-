```python
set=([4,3,2,1])
print(set)
```

```python
#adding an element

set={1,2,3,4}
set.add(5)
set
```

```python
#removing an element

set.remove(3)
set
```

```python
#discard()  ->safe from error

set.discard(5)
set
```

```python
#pop()

set.pop()
set
```

```python
#clear()

set.clear()
set
```

```python
#set operations
#union()

a={1,2,3,4,5,6}
r={5,6,7,8,9,10}
a.union(r)
```

```python
#intersection

a.intersection(r)
```

```python
#set difference (a-r)

a.difference(r)
```

```python
#set difference (r-a)

r.difference(a)
```

```python
#symmetric difference

a.symmetric_difference(r)
```
