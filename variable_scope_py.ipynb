```python
#local variable
def calculate_bill():
    amount=500   #calling only inside the function,can't modify
    print (amount)

calculate_bill()
```

```python
#global variable
tax=18           #calling outside the function (globally)
def show_tax():
    print(tax)

show_tax()
```

```python
"""count=10
def update():
    count=count+1 (ERROR)
update()"""


count=10
def update():
    global count
    count=count+1

update()
print(count)
```

```python
count=1
def increment():
    global count
    count=count+1  #to modify a variable firstly change the local variable to global by using the keyword global

increment()
print(count)
```
