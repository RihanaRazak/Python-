```python
#CLASS
class student:
    def __init__(self,name,roll,marks):
        self.name=name
        self.roll=roll
        self.marks=marks
    def display(self):
        print("name:",self.name)
        print("roll:",self.roll)
        print("marks:",self.marks,"\n")

#CREATING OBJECTS
s1=student("riya",150,90)
s2=student("nayra",151,95)

s1.display()
s2.display()
```

```python
class cat:
    def __init__(self,name,colour,gender):
        self.name=name
        self.colour=colour
        self.gender=gender
    def display(self):
        print("name:",self.name)
        print("colour:",self.colour)
        print("gender:",self.gender,"\n")

c1=cat("milky","white","female")
c2=cat("simba","white","male")
c3=cat("kitty","orange","female")
c4=cat("rockey","grey","male")
c5=cat("honey","black","female")

c1.display()
c2.display()
c3.display()
c4.display()
c5.display()
```

```python
#creating class (blueprint)

class car:
    def __init__(self,brand,colour): #init is the constructor
        self.brand=brand             #self refers to the current object
        self.colour=colour
    def start(self):
        print(self.brand,"car started")

#creating object (creating real life entity using class )

car1=car("BMW","white")
car2=car("baleno","silver")

print(car1.colour)
print(car2.brand)

car1.start()
car2.start()
```

```python
class BankAccount:
    def __init__(self,name,balance):
        self.name=name
        self.balance=balance

    def deposit(self,amount):
        self.balance+=amount
        print("deposited:",amount)

    def check_balance(self):
        print("balance:",self.balance)

acc1=BankAccount("rihana",5000)
acc2=BankAccount("nayra",10000)

acc1.deposit(2000)
acc1.check_balance()

acc2.deposit(2000)
acc2.check_balance()
```

```python
class mobile:

    def call(self):
        print("calling....")

    def message(self):
        print("sending messages....")

    def camera(self):
        print("taking photo....")

#creating object

m1=mobile()

#calling methods

m1.call()
m1.message()
m1.camera()
```

```python
class robot:

    def introduce(self):
        print("Hi I'm Siri, What can i do for you?")

    def task(self):
        print("I'm currently cleaning the room.")

#creating object

r1=robot()

#calling methods

r1.introduce()
r1.task()

```

```python
class theatre:
    def __init__(self,name,movie,seat):
        self.name=name
        self.movie=movie
        self.seat=seat

    def BookTicket(self):
        print(self.name ,"booked ticket for" ,self.movie,"\n")

    def ShowTicket(self):
        print("Details:")
        print("name:",self.name)
        print("seat no:",self.seat)
        print("movie:",self.movie,"\n")

    def CancelTicket(self):
        print(self.name,"cancelled ticket \n")



m1=theatre("rihana","jumanji",12)
m2=theatre("fidha","jurassic world",30)
m3=theatre("devika","titanic",20)

m1.BookTicket()
m1.ShowTicket()
m1.CancelTicket()

m2.BookTicket()
m2.ShowTicket()
m2.CancelTicket()

m3.BookTicket()
m3.ShowTicket()
```

```python
#inner class
class college:           #outer class
    class student:       #inner class
        def __init__(self,name,roll):
            self.name=name
            self.roll=roll
        def display(self):
            print("name:",self.name)
            print("roll_no:",self.roll)
s1=college.student("rahul",101)
s1.display()
```

```python
class company:
    def __init__(self,cname):
        self.cname=cname
    class employee:
        def __init__(self,ename):
            self.ename=ename
        def show(self):
            print("employee name:",self.ename)

c=company("TCS")
e=c.employee("anu")
e.show()
```

```python
class hospital:
    class doctor:
        def __init__(self,hname,name,specialist):
            self.hname=hname
            self.name=name
            self.specialist=specialist
        def display(self):
            print("hospital name:",self.hname)
            print("name:",self.name)
            print("specialization:",self.specialist)
s1=hospital.doctor("aster","Gopinadh","cardiologist")
s1.display()
```
