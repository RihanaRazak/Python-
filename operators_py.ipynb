```python
#arithmetic operators

r=10
a=20
print(r+a)
print(r-a)
print(r*a)
print(r/a)
print(r%a)
```

```python
#assignment operators

m=100
m+=10     #addition
print(m)
m-=10     #subtraction
print(m)
m*=10     #multiplication
print(m)
m/=10     #division
print(m)
m%=10     #percentage
print(m)
m**=2     #square
print(m)
m//=10    #floor division (keeps only the whole number part)
print(m)
```

```python
#Comparison/Relational operators

r=9
a=29

print(r>a)  #greater than
print(r<a)  #less than
print(r<=a) #less than or equal to
print(r>=a) #greater than or equal to
print(r==a) #equal to
print(r!=a) #not equal to
```

```python
#Logical operators

x=10
y=8
print (x==6 and y==8)
print (x==10 or y==8)
print (not y==0)
print (not x==10)
```

```python
#Bitwise operators

a=10       #binary=1010
b=4        #binary=0100
print(a&b) #and operator
print(a|b) #or operator
print(a^b) # XOR operator (when both values are same shows false)
print(~a)  # output=(-(a+1))
print(~b)
```

```python
# Identity operators

x=[1,2,3]  # (==)represent values (=)represent object
z=[1,2,3]
y=x
print(x is y )
print(x is not z)
```

```python
#Membership operators

q=[0,8,6,4]

print(0 in q)
print(9 in q)
print(2 not in q)
```

```python
#F-string/format-string

name="Rihana"
age=18
print(f"my name is {name} and i am {age} years old")
#or
print("my name is",name,"and i am",age,"years old")
```

```python
#swapping variables-pythonic way

a=10
b=20
a,b=b,a
print("a=",a,"b=",b)
```
